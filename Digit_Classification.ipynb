```python
#Splitting the data into training and validation sets
data = np.array(data)
#Risk of overfitting, risk of the model making predictions exactly suited for the training data, and not other data.
m, n = data.shape #The matrix shape, m = Number of columns, n = Number of rows
np.random.shuffle(data)

data_dev = data[0:1000].T #Transpose - First 1000 examples
Y_dev = data_dev[0] #Y_dev = The first row - The label
X_dev = data_dev[1:n] #X_dev = The rest of the rows
X_dev = X_dev / 255.

data_train = data[1000:m].T
Y_train = data_train[0]
X_train = data_train[1:n]
X_train = X_train / 255.
_,m_train = X_train.shape
```

```python
#Initializing all the parameters
def init_params():
  W1 = np.random.rand(10, 784) - 0.5 #Randn outputs between -0.5 and 0.5. (10, 784) - Dimensions of the array.
  b1 = np.random.rand(10, 1) - 0.5
  W2 = np.random.rand(10, 10) - 0.5
  b2 = np.random.rand(10, 1) - 0.5
  return W1, b1, W2, b2

def ReLU(Z):
  return np.maximum(Z, 0) # Z if Z>0   and   0 if X<=0

def softmax(Z):
  return np.exp(Z) / sum(np.exp(Z)) #np.exp(Z) is just like e^x


#Forward Propagation
def forward_prop(W1, b1, W2, b2, X):
  Z1 = W1.dot(X) + b1 #Dot Product of W1 and X + Bias.
  A1 = ReLU(Z1)
  Z2 = W2.dot(A1) + b2 #Dot Procuct of W2 and A1, W2 and the output layer of layer 0(Layer1) + Bias.
  A2 = softmax(Z2)
  return Z1, A1, Z2, A2


#Back Propagation

def deriv_ReLU(Z):
  return Z > 0

def one_hot(Y):
  one_hot_Y = np.zeros((Y.size, Y.max() + 1)) #Creates the correctly sized Matrix for us. Tuple of its size - Y.size = m (The number of examples), Y.max()+1 = Assumes the classes are 0 through 9, max is 9, then add 1 to get 10, the exact number of output classes that we want.
  one_hot_Y[np.arange(Y.size), Y] = 1 #Indexing through one_hot_Y using arrays, [np.arange(Y.size), Y] creates an array, range from 0 to m(To the number of training examples.), so [np.arange(Y.size), Y], specifies the row that we are accesing. Y specifies the column it accesses. Function: For each row, go to the column, specified by the label in Y and set it to one.
  one_hot_Y = one_hot_Y.T #Flip this, because right now each row is an example, and we want it the other way around, where each column is an example.
  return one_hot_Y

def back_prop(Z1, A1, Z2, A2, W1, W2, X, Y):
  #m = Y.size -Already defined before.
  one_hot_Y = one_hot(Y)
  dZ2 = A2 - one_hot_Y
  dW2 = 1 / m * dZ2.dot(A1.T)
  db2 = 1/ m * np.sum(dZ2)
  dZ1 = W2.T.dot(dZ2) * deriv_ReLU(Z1)

  dW1 = 1 / m * dZ1.dot(X.T)
  db1 = 1/ m * np.sum(dZ1)
  return dW1, db1, dW2, db2


#Update Parameters

def update_params(W1, b1, W2, b2, dW1, db1, dW2, db2, alpha):
  W1 = W1 - alpha * dW1
  b1 = b1 - alpha * db1
  W2 = W2 - alpha * dW2
  b2 = b2 - alpha * db2
  return W1, b1, W2, b2


#Gradient Descent

def get_predictions(A2):
  return np.argmax(A2, 0)

def get_accuracy(predictions, Y):
  print(predictions, Y)
  return np.sum(predictions == Y) / Y.size

def gradient_descent(X, Y, iterations, alpha):
  W1, b1, W2, b2 = init_params()
  for i in range(iterations):
    Z1, A1, Z2, A2 = forward_prop(W1, b1, W2, b2, X)
    dW1, db1, dW2, db2 = back_prop(Z1, A1, Z2, A2, W1, W2, X, Y)
    W1, b1, W2, b2 = update_params(W1, b1, W2, b2, dW1, db1, dW2, db2, alpha)
    if (i % 10 == 0):
      print("Iteration: ", i)
      print("Accuracy: ",  get_accuracy(get_predictions(A2), Y))
  return W1, b1, W2, b2
```

```python
W1, b1, W2, b2 = gradient_descent(X_train, Y_train, 500, 0.10)
```

```python
def make_predictions(X, W1, b1, W2, b2):
  _, _, _, A2 = forward_prop(W1, b1, W2, b2, X)
  predictions = get_predictions(A2)
  return predictions

def test_prediction(index, W1, b1, W2, b2):
  current_image = X_train[:, index, None]
  prediction = make_predictions(X_train[:, index, None], W1, b2, W2, b2)
  label = Y_train[index]
  print("Predictions: ", prediction)
  print("Label: ", label)

  current_image = current_image.reshape((28, 28)) * 255
  plt.gray()
  plt.imshow(current_image, interpolation = 'nearest')
  plt.show()
```
